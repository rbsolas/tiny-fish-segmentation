```python
from ultralytics import YOLO
```

```python
ROOT = "./drive/Shareddrives/[CS 198+199] Special Project"
MODEL = ROOT + "/models/yolov11"
DATA_YAML = "/datasets/usis10k/data.yaml"
```

```python
EPOCHS = 50
BATCH = 16
IMGSZ = 640 # CHANGE THIS
SAVE_PERIOD = 1
DEVICE = 0
PROJECT = "./"
NAME = DATA_YAML.split("/")[8] + f"_{EPOCHS}e_{BATCH}b"
SEED = 42
```

```python
# ----- ORIGINAL -----
# model_x = YOLO(MODEL + "/yolo11x-seg.yaml").load(MODEL + "/yolo11x-seg.pt")  # build from YAML and transfer weights

# ----- MODIFIED STRIDE -----
model_x_stride = YOLO(MODEL + "/yolo11x-seg_mod-stride.yaml").load(MODEL + "/yolo11x-seg.pt")  # build from YAML and transfer weights
```

```python
results = model_x_stride.train(data=DATA_YAML,
                        epochs=EPOCHS,
                        batch=BATCH,
                        imgsz=IMGSZ,
                        save_period=SAVE_PERIOD,
                        device=DEVICE,
                        project=PROJECT,
                        name=NAME,
                        seed=SEED)
```
