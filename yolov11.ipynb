```python
from ultralytics import YOLO
```

```python
model_x = YOLO("yolo11x-seg.yaml").load("yolo11x-seg.pt")  # build from YAML and transfer weights
```

```python
DATA_YAML = "../datasets/usis10k/data.yaml"

EPOCHS = 1
BATCH = 16
IMGSZ = 640
SAVE_PERIOD = 1
DEVICE = 0
PROJECT = "./"
NAME = DATA_YAML.split("/")[8] + f"_{EPOCHS}e_{BATCH}b"
SEED = 42
```

```python
results = model_x.train(data=DATA_YAML, 
                        epochs=EPOCHS, 
                        batch=BATCH, 
                        imgsz=IMGSZ, 
                        save_period=SAVE_PERIOD,
                        device=DEVICE,
                        project=PROJECT,
                        name=NAME,
                        seed=SEED)
```
